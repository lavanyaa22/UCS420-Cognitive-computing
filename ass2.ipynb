Generate a list of 100 random numbers between 100 and 900 (inclusive). Count and print all odd numbers in the list. Count and print all even numbers in the list. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
import random

random.seed(int(roll_number))

numbers = [random.randint(100, 900) for _ in range(100)]

odd_numbers = [x for x in numbers if x % 2 != 0]
print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))

even_numbers = [x for x in numbers if x % 2 == 0]
print("Even numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in numbers if is_prime(x)]

print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))

most_frequent = max(set(numbers), key=numbers.count)

print("Most frequent number:", most_frequent)
print("Occurrences:", numbers.count(most_frequent))
```

Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets. vi. vii. Find the union of A and B — all unique values across both sets. Find the intersection of A and B — values common to both. viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this differs from symmetric_difference(). ix. x. xi. Find the symmetric difference of A and B. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset(). Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain in one line why discard() is safer than remove() when you're not sure the value exists.

```python
digits = [int(digit) for digit in roll_number[:8]]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

print("Union:", A.union(B))
print("Intersection:", A.intersection(B))

print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

print("Symmetric Difference:", A.symmetric_difference(B))

print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))

x_val = input("Enter a value X: ")
if x_val.isdigit():
    X = int(x_val)
    A.discard(X)
    print("Set A after discard:", A)
else:
    print("Set A after discard:", A)
```
